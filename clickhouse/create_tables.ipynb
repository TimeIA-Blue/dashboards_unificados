```python
import clickhouse_connect
```

Script para criação das tabelas no **Clickhouse**, sincronizadas com o **SQLite** do gerenciador.

### Tabelas neste script

* ``params_radar``
* ``params_pedagio``
* ``params_vit``

```python
client = clickhouse_connect.get_client(
    host="52.44.163.66", username="default", password="defaultpass"
)
```

```python
client.query_df("SHOW TABLES")
```

```python
# dropando se já existir
client.query("DROP TABLE IF EXISTS params_radar")
client.query("DROP TABLE IF EXISTS params_pedagio")
client.query("DROP TABLE IF EXISTS params_vit")
```

```python
client.query_df("SHOW TABLES")
```

### Radar

```python
client.query(
    """
CREATE TABLE params_radar (
    `cnpj` String,
    `tipo_interrup` UInt8,
    `tempo_max_interrup` UInt16,
    `pct_max_interrup` Float32,
    `deploy_id` UInt16,
    `deploy_timestamp` DateTime
) ENGINE = MergeTree
ORDER BY
  (deploy_id, deploy_timestamp) SETTINGS index_granularity = 8192
"""
)
```

```python
client.query_df("SELECT * FROM params_radar")
```

### Pedágio

```python
client.query(
    """
CREATE TABLE params_pedagio (
    `cnpj` String,
    `tipo_interrup` UInt8,
    `tempo_max_interrup` UInt16,
    `deploy_id` UInt16,
    `deploy_timestamp` DateTime
) ENGINE = MergeTree
ORDER BY
  (deploy_id, deploy_timestamp) SETTINGS index_granularity = 8192
"""
)
```

### VIT

```python
client.query(
    """
CREATE TABLE params_vit (
    `cnpj` String,
    `vel_med` Float32,
    `soma_quatro_atrasos` UInt16,
    `it` UInt16,
    `deploy_id` UInt16,
    `deploy_timestamp` DateTime
) ENGINE = MergeTree
ORDER BY
  (deploy_id, deploy_timestamp) SETTINGS index_granularity = 8192
"""
)
```
